## Step 1: Prepare a Large Document

**What we do:**

- Count how many characters it has
- Count how many words it has

**Why?**

- Documents with more text show better compression results

````python
# STEP 1: Create a large document
print("=" * 70)
print("STEP 1: LARGE DOCUMENT")
print("=" * 70)

# This is our sample document about AI
document = """
Artificial Intelligence (AI) has become one of the most transformative technologies 
of the 21st century, fundamentally changing how we live and work. The field encompasses 
machine learning, deep learning, and natural language processing. These technologies 
power applications like Siri, Alexa, and Netflix recommendations.
"""

print(f"\n📄 Our Document:")
print(f"   '{document.strip()}'")
print(f"\n📊 Document Stats:")
print(f"   Characters: {len(document)}")
print(f"   Words: {len(document.split())}")
print(f"   Sentences: {len([s for s in document.split('.') if s.strip()])}")
````

## Step 2: Count Tokens (What Are Tokens?)

**What is a TOKEN?**
- A token = a small piece of text
- Usually 1 word = about 1.3 tokens
- Or 1 token = about 4 characters

**Example:**
- "Hello" = 1 token
- "Hello world" = 2 tokens
- "Artificial Intelligence" = 2-3 tokens

**Why count tokens?**
- AI models process text as tokens
- More tokens = more time and memory needed
- Compression = fewer tokens = faster & cheaper

````python
# STEP 2: Count tokens in our document
print("\n" + "=" * 70)
print("STEP 2: COUNT TOKENS BEFORE COMPRESSION")
print("=" * 70)

# Method 1: Using characters
# Rule: 1 token ≈ 4 characters
tokens_by_chars = len(document) / 4

# Method 2: Using words  
# Rule: 1 word ≈ 0.75 tokens (some words = multiple tokens)
words = len(document.split())
tokens_by_words = words * 0.75

# Average both methods
original_tokens = int((tokens_by_chars + tokens_by_words) / 2)

print(f"\n🧮 Token Counting Methods:")
print(f"   Method 1 (chars ÷ 4):     {int(tokens_by_chars)} tokens")
print(f"   Method 2 (words × 0.75):  {int(tokens_by_words)} tokens")
print(f"   Average:                  {original_tokens} tokens ✅")

print(f"\n💾 ORIGINAL DOCUMENT:")
print(f"   Tokens: {original_tokens}")
print(f"   Size: Large")
````

## Step 3: What Does CLaRa Do?

**CLaRa = Compression + Generation Model**

**Before Compression:**
```
"Artificial Intelligence (AI) has become one of the most transformative 
technologies of the 21st century, fundamentally changing how we live and work..."
→ 40+ tokens
```

**After Compression:**
```
"AI is a transformative technology that changes how we live and work..."
→ 15 tokens (4x smaller! 🚀)
```

**How does it work?**
1. Reads the whole document
2. Finds the IMPORTANT parts
3. Removes repetition and unnecessary words
4. Keeps the main meaning
5. Returns a much shorter version

````python
# STEP 3: Show compression examples
print("\n" + "=" * 70)
print("STEP 3: WHAT HAPPENS DURING COMPRESSION")
print("=" * 70)

print("\n🔄 COMPRESSION PROCESS:")
print("\n1️⃣  ORIGINAL TEXT:")
original_text = "Artificial Intelligence (AI) has become one of the most transformative technologies. AI is changing our lives. AI is everywhere."
print(f"   '{original_text}'")
print(f"   Tokens: ~{len(original_text.split()) * 0.75:.0f}")

print("\n2️⃣  CLaRa ANALYZES:")
print("   ✓ Identifies important info: AI, transformative, changing lives")
print("   ✓ Finds repetition: 'AI' mentioned 3 times")
print("   ✓ Removes redundancy: Keeps only needed info")

print("\n3️⃣  COMPRESSED OUTPUT:")
compressed_text = "Artificial Intelligence is a transformative technology changing our lives."
print(f"   '{compressed_text}'")
print(f"   Tokens: ~{len(compressed_text.split()) * 0.75:.0f}")
print(f"   Compression: {len(original_text.split()) / len(compressed_text.split()):.1f}x smaller! 🎉")

print("\n✅ MEANING PRESERVED:")
print("   Both texts say the SAME thing, but compressed version is much shorter")
````

## Step 4: Compression Ratios Explained

**Different Compression Levels:**

| Ratio | Original | Compressed | Space Saved |
|-------|----------|------------|-------------|
| **32x** | 500 tokens | 15 tokens | 97% smaller |
| **48x** | 500 tokens | 10 tokens | 98% smaller |
| **64x** | 500 tokens | 8 tokens | 98.4% smaller |

**What do these mean?**
- **32x**: Good balance between speed and quality
- **48x**: Fast, good results (recommended)
- **64x**: Super fast but might lose small details

````python
# STEP 4: Show different compression ratios
print("\n" + "=" * 70)
print("STEP 4: COMPRESSION RATIOS")
print("=" * 70)

original = 500  # 500 tokens

compression_ratios = [32, 48, 64]

print(f"\nSTARTING WITH: {original} tokens")
print("\nDIFFERENT COMPRESSION LEVELS:")
print("-" * 70)

for ratio in compression_ratios:
    compressed = original // ratio
    saved = original - compressed
    percentage = (saved / original) * 100
    
    print(f"\n🔹 {ratio}x Compression:")
    print(f"   Original:   {original} tokens")
    print(f"   Compressed: {compressed} tokens")
    print(f"   Saved:      {saved} tokens ({percentage:.1f}%)")
    print(f"   Speed boost: {ratio}x faster! 🚀")

print("\n💡 RECOMMENDATION: Use 48x for best balance")
print("   (Fast processing + keeps important details)")
````
